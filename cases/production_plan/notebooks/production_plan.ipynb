## 2. Linear model of the production plan

$F$ is the set of plants and $T$ the set of months. $c_f$ is the unit assembly cost of plant $f$ (€/bike), $K_{ft}$ its capacity in month $t$ (bikes/month) and $d_t$ the demand of month $t$ (bikes/month). The decision $p_{ft}$ is the number of bikes assembled in plant $f$ in month $t$, a continuous variable.

$$
\begin{aligned}
\min \quad & \sum_{f \in F} \sum_{t \in T} c_f\, p_{ft} && && \text{(assembly cost)} \\
\text{s.t.} \quad & \sum_{f \in F} p_{ft} \ge d_t && \forall t \in T && \text{(demand of each month)} \\
& p_{ft} \le K_{ft} && \forall f \in F,\ t \in T && \text{(capacity)} \\
& p_{ft} \ge 0 && \forall f \in F,\ t \in T
\end{aligned}
$$

The demand of a month is assembled in that month: the model has no stock and no orders served late.

## 3. Reduced version: plants 1 and 2 in January

With two plants and one month the model has two variables, $p_1$ and $p_2$, and can be drawn. Each constraint is a line: the capacity of plant 1 ($p_1 \le 250$), the capacity of plant 2 ($p_2 \le 200$) and the demand ($p_1 + p_2 \ge 380$). The points that meet all three are the **feasible region**. The dotted lines join the points of equal cost, $310\, p_1 + 340\, p_2$, one through each vertex; the optimum, in green, is the vertex on the lowest of them.

Below the figure, the vertices with their cost, and the **shadow price** of each constraint: the change in the optimal cost when its right-hand side grows by one bike. Each shadow price holds while the right-hand side stays in the range of the last column, found by solving the model again with the right-hand side changed.

The sliders change the capacity of plant 1 and the demand. Move the capacity of plant 1 below 180 and above 380: what happens to the optimum and to the shadow price of that capacity? What does a negative shadow price of a capacity mean?

```python
# @title Feasible region and shadow prices { display-mode: "form" }
capacity_plant_1 = 250  # @param {type:"slider", min:100, max:450, step:10}
demand_january = 380  # @param {type:"slider", min:200, max:500, step:10}
pp.feasible_region(capacity_plant_1, demand_january)
```
